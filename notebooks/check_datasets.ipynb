```python
import pandas as pd
import os
```

```python
pd.set_option("display.max_colwidth", None)
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)  # длинные таблицы (у HAR 562 колонки) обрезаются
pd.set_option('display.max_seq_items', None)
```

```python
datasets_folder = os.path.join('..', 'datasets')

# все файлы в папке, ниже проверяем каждый
sorted(os.listdir(datasets_folder))
```

```python
from IPython.display import display

def check(name):
    filepath = os.path.join(datasets_folder, name)
    data = pd.read_csv(filepath)
    print(f"file size: {os.path.getsize(filepath) / 1024:.0f} KB")
    print(f"rows: {len(data)}, columns: {len(data.columns)}")
    print(data.columns)
    # для каждой колонки: тип, сколько разных значений и сколько пропусков
    display(pd.DataFrame({'type': data.dtypes, 'unique': data.nunique(), 'missing': data.isna().sum()}))
    display(data.describe().T)
    return data
```

```python
# пропуски только в 11 колонках, больше всего в PoolQC, MiscFeature, Alley и Fence (пустых больше 80%)
# цена дома в SalePrice
data = check('Ames_Housing_Sales.csv')
data.head(10).T
```

```python
# 561 числовой признак уже нормирован в [-1, 1], цель это Activity (6 видов активности)
data = check('Human_Activity_Recognition_Using_Smartphones_Data.csv')
data.head(10).T
```
